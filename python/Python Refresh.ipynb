### Python slicing and indexing

```python
lst = [1, 2, 3, 5, 9]
print(lst[-1])
print(lst[-2])
```

```python
# 1D slicing
lst = list(range(10))
print(lst[2:5])
print(lst[:4]) # start omitted
print(lst[5:]) # stop omitted
print(lst[::2]) # both start and stop omitted, step=2
print(lst[::-1]) # reverse list
```

```python
# 2D slicing
import numpy as np
arr = np.arange(20).reshape(4,5)
arr
```

```python
print(arr[:, -3:]) # last 3 columns of every row
```

```python
print(arr[:2, 1:4])
```

```python
print(arr[2,3])
```

```python
arr[2,3] = 99 # in numpy array, slicing returns a view, so changing the slice can change the original array
arr
```

```python
arr[::2, 1:3]
```
